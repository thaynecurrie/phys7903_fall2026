### If-Then Statements

#### _Basic Examples_

If-Then statements perform an operation if a statement is True.   "If [x is true], then do [y]".   The if-then statements are ended by a ```:```.   

The program ``ex29.py`` is an example:

```
def execute(people=20,cats=30,dogs=15):
 if people < cats:
    print("Too many cats! The world is doomed!")

 if people > cats:
    print("Not many cats! The world is saved!")

 if people < dogs:
    print("The world is drooled on!")

 if people > dogs:
    print("The world is dry!")


 dogs += 5

 if people >= dogs:
    print("People are greater than or equal to dogs.")

 if people <= dogs:
    print("People are less than or equal to dogs.")


 if people == dogs:
    print("People are dogs.")

```

Before we proceed, note one thing about the values for variables "people", "cats", and "dogs" in the function call.  They seem to be assigned specific values.   And indeed the values listed are exactly what are used if you do _not_ override them at the function call.

So what happens in this program?  Depending on the value of "people", "dogs", and "cats" different things are printed.   Here are some examples:

#### _If/El-If/Else  Statement Structures_

Now we are going to add another wrinkle: If/El-If/Else structures.  Before, with an if-then statement, we execute a block of code if a given condition is met.   The If/El-If/Else is like that but allows for a branching of code.

See _ex30.py_ below

```
def execute(people=30,cars=40,trucks=15):
 if cars > people:
    print("We should take the cars.")
 elif cars < people:
    print("We should not take the cars.")
 else:
    print("We can't decide.")

 if trucks > cars:
    print("That's too many trucks.")
 elif trucks < cars:
    print("Maybe we could take the trucks.")
 else:
    print("We still can't decide.")

 if people > trucks:
    print("Alright, let's just take the trucks.")
 else:
    print("Fine, let's stay home then.")

```

With no changes to the variables, this executes as ...

```
>> from ex30 import execute
>> execute()

We should take the cars.
Maybe we could take the trucks.
Alright, let's just take the trucks.


```

or, setting cars =0, we get ...

```
>>> execute(cars=0)
We should not take the cars.
That's too many trucks.
Alright, let's just take the trucks.
```



### Logic and Truth Values

[shamelessly copied from Learning Python the Hard Way]

Logic describes whether things are or are-not, true or not-true (false), etc.  They form the backbone of decision trees in code.   

Python uses different key _terms_ to decide whether or not something is True or False.   Below is a list of truth _terms_.   Logic on a computer is about seeing whether some combination of these _terms_ is True at a given point in the program:

* ```and```
* ```or```
* ```not```
* ```!=``` (not equal)
* ```==``` (equal)
* ```>=``` (greater than or equal to)
* ```<=``` (less than or equal to)
* ```True``` 
* ```False```

***Important***: the ``None`` logic value

One practical application of this is the ``not None`` syntax.  You will see that in a lot of Python.  E.g.  A function has several default values 

```
def atest(first_var=None,second_var=100,third_var=200):
[stuff]

 if first_var is not None:
   #then do stuff

 else:
  do something else

```

Since I was 'raised' on IDL, I treat the ``if [keyword] is not None`` conditional as similar to ``if keyword_set([variable name])`` in IDL.

### Using Logic Values to Create Compound Conditionals

## Logic and Truth Values: for Scalars vs. Arrays

Here's a "gotcha".  In all the cases above -- ``and``, ``or``, ``None`` -- the conditionals are focused on _scalars_ with _single_ True/False values.  

E.g. 
```
variable=10
variable2=8
if variable > 5 and variable2 > 4: #variable > 5 is a single truth value, variable2 > 4 is also a single truth value
 [do something]
```

or

```
variable=10
variable2=8
if variable > 5 and variable < 12: #variable > 5 is a single truth value, variable < 12 is also a single truth value
 [do something]
```

In real Python code, you will often instead see things that look like this:

```
variable=np.array([1,2,3,5,10])
good = (variable > 2) & (variable < 10)
#or
good=np.where((variable > 2) & (variable < 10) )
#or
variable_new=variable[ (variable > 2) & (variable < 10) ]
#or
good = (variable > 2) & (variable < 10)
variable_new=variable[ good ]
```

Another example

```
alsogood = (variable > 5) | (variable <3 )
#or
alsogood=np.where((variable > 5) | (variable <3 ) )
#or
alsovariable_new=variable[ (variable > 5) | (variable <3 ) ]
#or
alsogood = (variable > 5) | (variable <3 )
variable_new=variable[ alsogood ]
```

Here, the ``&`` symbol means "and" and ``|`` means "or".  *However*, they work for *arrays*, not single-valued scalars.  I.e. the above code block will work -- and ``and``/``or`` will not -- if ``variable`` has more elements than just a single number.

Watch what happens when we try to do the following:

````python
import numpy as np 
arr = np.array([2,3,5,6,8,9,10]) #defines an array, more on this later

# 'and' on arrays FAILS -- a comparison gives a whole array of True/False,
# and 'and' can't collapse that into one answer:

#don't worry about the try/except block yet.  You'll see it everywhere: we will cover it. This is basically another conditional
try:
    result = arr[(arr > 3) and (arr < 9)]
except ValueError as e:
    print("FAILS:", e)

# use & (and), | (or), ~ (not) instead -- they work element-by-element:
print("works:", arr[(arr > 3) & (arr < 9)])   # -> [5 6 8]
````

What the code just told us is that ``arr`` has an array of values (7 to be exact).  

Then we try to do a conditional with ``and``.  This fails because ``and`` (and ``or`` and ``not``) only works with a SINGLE True/False value.  The alternate syntax shown above (labeled "works") returns _all_ of the True cases for the array.

**tl;dr:**

If you are doing an ``if/elif/else`` block and need to evaluate a single True/False value, use ``and``/``or``/``not``.  If you have an array (again, we will get to this later), you need to use different syntax (``&``, ``|``, etc.).
